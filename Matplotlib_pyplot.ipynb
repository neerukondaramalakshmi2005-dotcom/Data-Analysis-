```python
# What is Matplotlib?
Matplotlib is a Python library used to create data visualizations and graphs.
It is one of the most commonly used visualization libraries in Python, especially for Data Analysis, Data Science, Machine Learning, and Exploratory Data Analysis (EDA).
Matplotlib can be used to create:
Bar charts
Horizontal bar charts
Histograms
Line plots
Multiple line charts
Scatter plots
Pie charts
Donut charts
And many other customized visualizations
Why do we use Matplotlib?
Raw data can be difficult to understand.
For example:
sales = [10000, 15000, 12000, 18000, 22000]
Looking at these numbers doesn't immediately show the trend.
A graph can make it easier to identify:
Highest sales
Lowest sales
Increasing/decreasing trends
Relationships between variables
Distribution of data
Category-wise comparisons
Simple idea
Raw Data
   ↓
Matplotlib
   ↓
Visualization
   ↓
Easy Analysis

# 14. Why Learn Matplotlib Before Seaborn and Plotly?
Matplotlib provides the foundation of Python visualization.
Students learn concepts such as:
X-axis
Y-axis
Figure
Axes
Labels
Titles
Legend
Markers
Lines
Bars
Colors
Annotations









```

```python

#1. Create Simple Dataset

# We will use a small Product Sales dataset
#products → Categorie #sales → Sales values #profit → Profit values

import matplotlib.pyplot as plt
import numpy as np
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.plot(products,sales,marker="*")
plt.title("products,sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
#2. Line Chart (Purpose Used to show trends or changes.)

plt.plot(products, sales, marker="o")

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# 3. Multiple Line Chart (Used to compare two or more trends.)

# LINE CHART
# IT IS MULTIPLE  DATA from matplotlib import legend
import matplotlib.pyplot as plt

products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.plot(products,sales,marker="*",label="sales")
plt.plot(products,profit,marker="*",label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python

# 4. BAR CHART ( Used to compare values between categories)

import matplotlib.pyplot as plt
import numpy as np
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.bar(products,sales,label="sales")
plt.title("products,sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python

# 5. Horizontal Bar Chart (Same information as a bar chart, but displayed horizontally.)

import matplotlib.pyplot as plt
import numpy as np
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.barh(sales,products,label="sales")
plt.title("products,sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python

# 6. Grouped Bar Chart (Used to compare Sales and Profit for every product.)

x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2, sales, width, label="Sales")
plt.bar(x + width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python

#7. Stacked Bar Chart (Used to show how different values contribute to a total.)

# stacked bar chart ( it is the symbols of + , - ) ++ , -- = up's and down same
x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2, sales, width, label="Sales")
plt.bar(x - width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python

#10. Pie Chart
#Used to show parts of a whole.
import matplotlib.pyplot as plt
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
#11. Donut Chart
#A donut chart is basically a pie chart with a hollow center.
import matplotlib.pyplot as plt
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.title("Product Sales Distribution")

plt.show()
```

```python
#12. Donut Chart with Center Text
#Now we can put information inside the hollow area.
import matplotlib.pyplot as plt

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.text(
    0,
    0,
    "Total\n650",
    ha="center",
    va="center"
)

plt.title("Total Product Sales")

plt.show()
```

```python
#13. Area Chart
#Matplotlib does not have a function called area_chart(). We create an area chart using fill_between().
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [100, 120, 150, 140, 180, 200]

plt.plot(months, monthly_sales)
plt.fill_between(months, monthly_sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#8. Histogram
#A histogram is used to understand the distribution of numerical data.
#Let's create student marks.
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 65, 68, 70,
         72, 75, 75, 78, 80, 82, 85, 88, 90, 95]
plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Stake Bar chart
# it is also  same like area chart (one area of measure) and
# it is  also the multiple measures  in a single categorie (months) chart
import matplotlib.pyplot as plt

months = [ "Jan", "Feb", "Mar", "Apr","May","Jun"]
laptop = [40, 50, 60, 70, 80 ,90]
mobile = [ 60, 70, 80, 85,100,120]
plt.stackplot(
    months,laptop,mobile,
    labels=["laptop","mobile"]
)
plt.title("laptop and mobile sales")
plt.xlabel("month")
plt.legend()
plt.show()
```

```python
# 16. Multiple Box Plot
# We can compare the distributions of multiple groups.
import matplotlib.pyplot as plt

python_marks = [60, 65, 70, 72, 75, 80, 85]
sql_marks = [55, 60, 65, 70, 72, 78, 82]
powerbi_marks = [50, 60, 68, 72, 75, 80, 90]

plt.boxplot(
    [python_marks, sql_marks, powerbi_marks],
    labels=["Python", "SQL", "Power BI"]
)

plt.title("Student Performance")

plt.ylabel("Marks")

plt.show()
```

```python
# scatter plot
# it is used for the relationship between numerical variables
# it derives the depedendency and un depedendency
#Used to find the relationship between two numerical variables.
#Let's assume:
#Advertising expenditure
#Sales
import matplotlib.pyplot as plt

advertising = [10, 20, 30, 40, 50, 60, 70, 80]
sales_amount = [25, 35, 40, 50, 55, 65, 72, 85]

plt.scatter(advertising, sales_amount)
plt.title("Advertising vs Sales")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.show()
```
