```python
# 3. Multiple Line Chart (Used to compare two or more trends.)

# LINE CHART
# IT IS MULTIPLE  DATA from matplotlib import legend
import matplotlib.pyplot as plt

products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.plot(products,sales,marker="*",label="sales")
plt.plot(products,profit,marker="*",label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
#plt.legend()
plt.show()
```

```python

# 4. BAR CHART ( Used to compare values between categories)

import matplotlib.pyplot as plt
import numpy as np
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.bar(products,sales,label="sales")
plt.title("products,sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python

# 5. Horizontal Bar Chart (Same information as a bar chart, but displayed horizontally.)

import matplotlib.pyplot as plt
import numpy as np
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.barh(sales,products,label="sales")
plt.title("products,sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python

# 6. Grouped Bar Chart (Used to compare Sales and Profit for every product.)

x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2, sales, width, label="Sales")
plt.bar(x + width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python

#7. Stacked Bar Chart (Used to show how different values contribute to a total.)

# stacked bar chart ( it is the symbols of + , - ) ++ , -- = up's and down same
x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2, sales, width, label="Sales")
plt.bar(x - width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python

#10. Pie Chart
#Used to show parts of a whole.
import matplotlib.pyplot as plt
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
#11. Donut Chart
#A donut chart is basically a pie chart with a hollow center.
import matplotlib.pyplot as plt
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.title("Product Sales Distribution")

plt.show()
```

```python
#12. Donut Chart with Center Text
#Now we can put information inside the hollow area.
import matplotlib.pyplot as plt
products = [" laptop","mobile","Tablet","Headphones","Watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.text(
    0,
    0,
    "Total\n650",
    ha="center",
    va="center"
)

plt.title("Total Product Sales")

plt.show()
```

```python
#13. Area Chart
#Matplotlib does not have a function called area_chart(). We create an area chart using fill_between().
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [100, 120, 150, 140, 180, 200]

plt.plot(months, monthly_sales)
plt.fill_between(months, monthly_sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#8. Histogram
#A histogram is used to understand the distribution of numerical data.
#Let's create student marks.
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 65, 68, 70,
         72, 75, 75, 78, 80, 82, 85, 88, 90, 95]
plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Staked Bar chart
# it is also  same like area chart (one area of measure) and
# it is  also the multiple measures  in a single categorie (months) chart
import matplotlib.pyplot as plt

months = [ "Jan", "Feb", "Mar", "Apr","May","Jun"]
laptop = [40, 50, 60, 70, 80 ,90]
mobile = [ 60, 70, 80, 85,100,120]
plt.stackplot(
    months,laptop,mobile,
    labels=["laptop","mobile"]
)
plt.title("laptop and mobile sales")
plt.xlabel("month")
plt.legend()
plt.show()
```

```python
# contour plot( It is 3-Dimensional but represented the 2-Dimensioanal surface )
# it is a create a small numerical  values,same equal value)
# it is Graphical map represented.
# it using the meshGrid( x,y,z(z -3D to 2D))
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1, 11)
y = np.arange(1, 11)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2
plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# 3D line plot
# it is multiple data points with lines in three-dimensional spaces
# graph contains x,y,z.
# trends,movement,
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
z = [5, 10, 15, 20, 25]

ax.plot(x, y, z)

ax.set_title("3D Line Plot")

plt.show()
```

```python
# 3D scatter plot
# it is individual data points in three-dimensional space.
# three numerical data (ax.scatter (x,y,z)

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]
z = [2, 4, 8, 16, 32]

ax.scatter(x, y, z)

ax.set_title("3D Scatter Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```

```python

#3D Bar Chart

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4]
y = [1, 1, 1, 1]
z = [10, 20, 15, 25]

dx = [0.5, 0.5, 0.5, 0.5]
dy = [0.5, 0.5, 0.5, 0.5]
dz = z
ax.bar3d(x, y, z, dx, dy, dz)
ax.set_title("3D Bar Chart")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
