```python

import seaborn as sns
import matplotlib.pyplot as plt
d = sns.load_dataset("tips")
sns.histplot(data=d,x="total_bill",hue="smoker",bins =10)
plt.title("Total Bill Distribution")
plt.show()
print(d)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
d = sns.load_dataset("tips")
sns.scatterplot(data=d,x="total_bill",y="tip")
plt.title("Total Bill vs Tip")
plt.show()
print(d)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
d = sns.load_dataset("tips")
sns.boxplot(data=d,x="day",y="total_bill")
plt.title(" Bill Distribution by Day ")
plt.show()
print(d)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
d = sns.load_dataset("tips")
sns.violinplot(data=d,x="total_bill",y="tip")
plt.title("Total Bill vs Tip")
plt.show()
print(d)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
d = sns.load_dataset("tips")
sns.lineplot(data=d,x="total_bill",y="tip",hue="sex",marker="*",estimator="mean")
plt.title("Total Bill vs Tip")
plt.show()
print(d)
```
