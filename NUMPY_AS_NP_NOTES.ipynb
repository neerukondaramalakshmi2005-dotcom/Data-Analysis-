```python
# Aggcal :- it is mathemitaical or statical
#SUM,MIN,MAX,ME
import numpy as np
#============ sum 1D ================
arr = np.array([10,20,40,30,50])
print(np.sum(arr))
#=============sum 2D =================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.sum(arr))
# ============ row wise sum 2D ================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.sum(arr,axis=0))
# ============ row wise sum 2D ================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.sum(arr,axis=1))
#============ min 1D ================
arr = np.array([10,20,40,30,50])
print(np.min(arr))
# ==============min 2D==============
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.min(arr))
#============ max 1D ================
arr = np.array([10,20,40,30,50])
print(np.max(arr))
#============ max 2D===================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.max(arr))
#============mean 1D====================
arr = np.array([10,20,40,30,50])
print(np.mean(arr))
#============ mean 2D===================( sum of values and number of values)
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.mean(arr))
#============std 1D====================( over all  differ from the mean)
arr = np.array([10,20,40,30,50])
print(np.std(arr))
#============ std 2D===================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.std(arr))
#============var 1D====================( it is varience) spread out of mean
arr = np.array([10,20,40,30,50])
print(np.var(arr))
#============ var 2D===================Low variance → values are close together
#                                     High variance → values are widely spread

arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.var(arr))
# ================percentile 1D ()==========(below which a certain percentage of observations fall.)
# it is only for the (1 to 100) The 50th percentile is the median.
#25th percentile → Q1
#50th percentile → Median
#75th percentile → Q3

arr = np.array([10,20,40,30,50])
print(np.percentile(arr,30))
#=============== percentile 2D========================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.percentile(arr,100))

# percentail and quantail ( both are vise versa)
#percentile → percentage → 0–100
#quantile    → proportion → 0–1

# ================== quantile ====================( it only taken for 0 to 1 values only)
arr = np.array([10,20,30,40,50,80])
print(np.percentile(arr,25))
print(np.quantile(arr,0.25))
#============= argmax()======================( it is index position of the largest value
arr = np.array([10,20,40,30,50])
print(np.argmax(arr))
#============ argmax() 2D===================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.argmax(arr))
#============= argmin()======================
arr = np.array([10,20,40,30,50])
print(np.argmin(arr))
#============ argmin() 2D===================
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
# =================log10===================
arr = np.array([10,20,40,30,50])
print(np.log10(arr))
# =================log10===================
arr = np.array([10,100,1000,100000])
print(np.log10(arr))
print("_________________________________- ")
print(np.log2(arr))


# ==============log10 2D==============
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.log10(arr))
print(" ______________________________  ")
arr = np.array([10,20,40,30,50,60]).reshape(2,3)
print(np.log(2))
print(np.log10(arr))
print(np.log(3))
#===================== Round()===================( it is only 5 values=)
arr = np.array([10.555, 20.786,30.676]) # ( it is only taken for Decimal values like 10.20 taken the value of 10)
print(np.round(arr,2))
# ==================floor=================( it include the lower value only)
arr = np.array([10.2, 20.4,30.5])
print(np.floor(arr))
# ================== ceil===================( it is upper value only) like we 10.2 = 11.)
arr = np.array([10.2, 20.3, 30.4, 50.4])
print(np.ceil(arr))
#=================== abs() absoult method it is used to negetaive also convert the positive value
# abs is making into positive
arr = np.array([-10, -4, -70,0,8,90])
print(np.abs(arr))
#================= sort()=============== (it is descending [::-1] order to change)
arr = np.array([50,40,60,20,60])
result = np.sort(arr)[::-1]
print(result)
#================sort()=====================
arr = np.array([10,20,40,30,50])
print(np.sort(arr))
```

```python
# concatenate() combining  multiple arrays and to or more arrays  to result would be same dimension
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])
print(arr1,arr2)
result = np.concatenate((arr1,arr2))
print(result.ndim)
print(arr1.shape)
print(arr2.shape)
print("_______________________________")
#============================================
# axis =0 means colums wise concatenadation
#===============2D array ====================
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.concatenate((arr1,arr2),axis =0)
print(result)
print(result.ndim)
print("_________________________________")

#================1D axis = 0 =====================
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.concatenate((arr1,arr2),axis =0)
print(result)
print("_________________________________")

#===============2D array axis =1  ====================
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.concatenate((arr1,arr2),axis =1)
print(result)
print(result.ndim)
print("_________________________________")
#===============1D array axis =1  ====================
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.concatenate((arr1,arr2),axis =1)
print(result)

print("_________________________________")
#========stack axis =0 ============================
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.stack((arr1, arr2),axis= 0)
print(result)
#========stack axis =0 ============================
print("____________________________-")
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.stack((arr1, arr2),axis= 0)
print(result)
print(result.ndim)
#========stack axis =1============================
print("______________________________")
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.stack((arr1, arr2))
print(result)
#========stack axis =1============================
print("_________________________----")
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.stack((arr1, arr2),axis= 1)
print(result)
print(result.ndim)
# v stack () it is vertical stacking (high and down)
# ============ 1D=================
print("v1d_____________________________--")
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.vstack((arr1, arr2))
print(result)

#============= exceptional case of vstack of 2D============
print("v2d_____________________________")
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.vstack((arr1,arr2))
print(result)
#============= exceptional case of vstack of 3D============
print("v3d_____________________________")
arr1 = np.array([[10,20],[30,4],[80,50]])
arr2 = np.array([[50,60],[70,80],[40,60]])
arr3 = np.array([[100,110],[120,130],[140,150]])
result = np.vstack((arr1,arr2,arr3))
print(result)
# Horizontal stack(hstack) it is side by side
# ============ hstack of  1D=================
print("h1d_____________________________--")
arr1 = np.array([10,20,30])
arr2 = np.array([50,60,70])
result = np.hstack((arr1, arr2))
print(result)
print(result.ndim)

#============= exceptional case of hstack of 2D============
print("h2d_____________________________")
arr1 = np.array([[10,20],[30,4]])
arr2 = np.array([[50,60],[70,80]])
result = np.hstack((arr1,arr2))
print(result)
print(result.ndim)
#============= exceptional case of hstack of 3D============
print("h3d_____________________________")
arr1 = np.array([[10,20],[30,4],[80,50]])
arr2 = np.array([[50,60],[70,80],[40,60]])
arr3 = np.array([[100,110],[120,130],[140,150]])
result = np.hstack((arr1,arr2,arr3))
print(result)
print(result.ndim)

```
