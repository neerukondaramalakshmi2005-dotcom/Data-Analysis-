```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample employee data
employees = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "Arun", "Divya", "Kiran"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Age": [24, 29, 32, 27, 35, 30],
    "Salary": [45000, 38000, 60000, 42000, 50000, 55000]
})

# 1. Department-wise employee count and average salary
department_summary = employees.groupby("Department").agg(
    Employee_Count=("Name", "count"),
    Average_Salary=("Salary", "mean")
)
print("Department summary:")
print(department_summary)

# 2. Department with the highest average salary
highest_average_department = department_summary["Average_Salary"].idxmax()
print("\nDepartment with the highest average salary:",
      highest_average_department)

# Department with the minimum individual salary
minimum_salary_employee = employees.loc[employees["Salary"].idxmin()]
print("\nEmployee with the minimum salary:")
print(minimum_salary_employee[["Name", "Department", "Salary"]])

# 3. Chart comparing salary distributions across departments
employees.boxplot(column="Salary", by="Department")
plt.title("Salary Distribution by Department")
plt.suptitle("")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()

# Chart showing the relationship between age and salary
plt.scatter(employees["Age"], employees["Salary"])
plt.title("Relationship Between Age and Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.tight_layout()
plt.show()

# 4. Create 30 days of sales data and plot it
np.random.seed(10)  # Makes the random results repeatable

dates = pd.date_range(start="2026-01-01", periods=30, freq="D")
daily_sales = np.random.randint(1000, 10000, size=30)

sales_30_days = pd.DataFrame({
    "Date": dates,
    "Sales": daily_sales
})

print("\n30 days of sales data:")
print(sales_30_days)

plt.plot(sales_30_days["Date"], sales_30_days["Sales"], marker="o")
plt.title("Daily Sales for 30 Days")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 5. Generate random sales values for 20 products
product_sales = np.random.randint(1000, 10001, size=20)

product_sales_df = pd.DataFrame({
    "Product": [f"Product {i}" for i in range(1, 21)],
    "Sales": product_sales
})

print("\nRandom sales for 20 products:")
print(product_sales_df)

# 6. Monthly sales trend
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [25000, 29000, 30000, 40000, 45000, 52000]

monthly_sales_df = pd.DataFrame({
    "Month": months,
    "Sales": monthly_sales
})

print("\nMonthly sales data:")
print(monthly_sales_df)

plt.plot(monthly_sales_df["Month"], monthly_sales_df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```
